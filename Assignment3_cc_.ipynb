```python
# Q6 - Create employee dataset

employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": [
        "2020-03-15",
        "2017-07-19",
        "2013-06-01",
        "2021-02-10",
        "2010-11-25"
    ],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(employee_data)

employees.to_csv("employees.csv", index=False)

print(employees)
```

```python
# a) Shape

print("Shape of DataFrame:", employees.shape)
```

```python
# b) Summary

print(employees.info())
```

```python
# c) Descriptive statistics

print(employees.describe())
```

```python
# d) First 5 rows

print("First 5 rows:")
print(employees.head(5))

print("\nLast 3 rows:")
print(employees.tail(3))
```

```python
# e) Statistics

average_salary = employees["Salary"].mean()
total_bonus = employees["Bonus"].sum()
youngest_age = employees["Age"].min()
highest_rating = employees["Rating"].max()

print("Average salary:", average_salary)
print("Total bonus:", total_bonus)
print("Youngest employee's age:", youngest_age)
print("Highest performance rating:", highest_rating)
```

```python
# f) Sort by Salary in descending order

sorted_employees = employees.sort_values(
    by="Salary",
    ascending=False
)

print(sorted_employees)
```

```python
# g) Add Performance Category

def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance"] = employees["Rating"].apply(
    performance_category
)

print(employees)
```

```python
# h) Missing values

print(employees.isnull().sum())
```

```python
# i) Rename Employee_ID to ID

employees.rename(
    columns={"Employee_ID": "ID"},
    inplace=True
)

print(employees)
```

```python
# j.1) Employees with more than 5 years of experience

more_than_5 = employees[
    employees["Years_of_Experience"] > 5
]

print("Employees with more than 5 years of experience:")
print(more_than_5)
# j.2) Employees belonging to IT department

it_employees = employees[
    employees["Department"] == "IT"
]

print("Employees in IT department:")
print(it_employees)
```

```python
# k) Add Tax column

employees["Tax"] = employees["Salary"] * 0.10

print(employees)
```

```python
# l) Save modified DataFrame

employees.to_csv(
    "modified_employees.csv",
    index=False
)

print("Modified DataFrame saved as modified_employees.csv")
```
