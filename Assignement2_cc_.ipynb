```python
# Q1
roll_no = 1024170417

# Extract individual digits and multiply each digit by 10
digits = [int(digit) for digit in str(roll_no)]
L = [digit * 10 for digit in digits]

# i. Print L
print("Initial L:", L)

# ii. Add two numbers using append() and insert()

L.append(50)
print("After append(50):", L)
# append() adds 50 at the end of the list.

L.insert(2, 30)
print("After insert(2, 30):", L)
# insert() adds 30 at index 2 and shifts the existing elements to the right.

# iii. Remove two elements using remove() and pop()

L.remove(10)
print("After remove(10):", L)
# remove() deletes the first occurrence of the specified value 10.

L.pop(1)
print("After pop(1):", L)
# pop(1) removes and returns the element at index 1.

# iv. Sort in ascending and descending order

L.sort()
print("Ascending order:", L)

L.sort(reverse=True)
print("Descending order:", L)

# v. Slicing: first three and last three elements

print("First three elements:", L[:3])
print("Last three elements:", L[-3:])

# vi. List comprehension: elements greater than average

average = sum(L) / len(L)

greater_than_average = [x for x in L if x > average]

print("Average:", average)
print("Elements greater than average:", greater_than_average)
```

```python
# Q2
# Use the first 8 values from the original L
# (before the append/insert/remove operations)

digits = [int(digit) for digit in str(1024170417)]
original_L = [digit * 10 for digit in digits]

scores = tuple(original_L[:8])

print("Scores tuple:", scores)

# i. Highest score and its index
highest = max(scores)
highest_index = scores.index(highest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)

# Lowest score and number of times it appears
lowest = min(scores)
lowest_count = scores.count(lowest)

print("Lowest score:", lowest)
print("Number of times lowest score appears:", lowest_count)

# ii. Reverse the tuple and return it as a list
# Tuples are immutable, so they cannot be reversed in place.
reversed_scores = list(reversed(scores))

print("Reversed tuple as list:", reversed_scores)

# iii. Ask the user for a score
user_score = int(input("Enter a score: "))

if user_score in scores:
    print("First occurrence index:", scores.index(user_score))
else:
    print("Score not present in tuple")

# iv. Attempt to change an element
# Tuples are immutable, so their elements cannot be changed like list elements.

try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
    print("Reason: Tuples are immutable, whereas list elements can be modified.")

# v. Tuple unpacking using *
first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
# Q3
import random

# Set seed to your roll number
random.seed(1024170417)

# i. Generate 100 random numbers between 100 and 900
numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(numbers)

# ii. Count and print all odd numbers

odd_numbers = [x for x in numbers if x % 2 != 0]

print("\nOdd numbers:")
print(odd_numbers)
print("Count of odd numbers:", len(odd_numbers))

# iii. Count and print all even numbers

even_numbers = [x for x in numbers if x % 2 == 0]

print("\nEven numbers:")
print(even_numbers)
print("Count of even numbers:", len(even_numbers))

# iv. Count and print all prime numbers

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


prime_numbers = [x for x in numbers if is_prime(x)]

print("\nPrime numbers:")
print(prime_numbers)
print("Count of prime numbers:", len(prime_numbers))

# v. Find the number occurring most frequently

frequency = {}

for x in numbers:
    frequency[x] = frequency.get(x, 0) + 1

most_frequent = max(frequency, key=frequency.get)

print("\nMost frequently occurring number:", most_frequent)
print("Number of times it occurs:", frequency[most_frequent])
```

```python
# Q4

# Extract the first 8 digits of your roll number
digits = [int(digit) for digit in str(1024170417)[:8]]

# Create sets A and B
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

# vi. Union
union = A.union(B)
print("\nUnion of A and B:", union)

# vii. Intersection
intersection = A.intersection(B)
print("Intersection of A and B:", intersection)

# viii. A - B and B - A
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference() gives elements present in one set but not the other.
# symmetric_difference() gives all elements that are not common to both sets.

# ix. Symmetric difference
symmetric_diff = A.symmetric_difference(B)

print("Symmetric difference:", symmetric_diff)

# x. Check subset and superset

print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# xi. Ask user for a value and remove it using discard()

X = int(input("\nEnter a value to remove from A: "))

A.discard(X)

print("Set A after discard:", A)

# discard() is safer than remove() because discard() does not give an error
# if the specified value is not present in the set.
```

```python
# Q5

my_dict = {
    "name": "Anvi",
    "roll_no": "1024170417",
    "branch": "COPC",
    "age": 19,
    "city": "Kaithal"
}

print("Original dictionary:")
print(my_dict)

# i. Rename "city" to "location" without recreating the dictionary

my_dict["location"] = my_dict.pop("city")

print("\nAfter renaming city to location:")
print(my_dict)

# ii. Add CGPA

my_dict["cgpa"] = 8.5

print("\nAfter adding CGPA:")
print(my_dict)

# iii. Increase age by 1

my_dict["age"] += 1

print("\nAfter increasing age by 1:")
print(my_dict)

# iv. Delete branch using pop() and del in two separate copies

dict_pop = my_dict.copy()
dict_del = my_dict.copy()

# Using pop()
removed_branch = dict_pop.pop("branch")

print("\nUsing pop():")
print(dict_pop)
print("Removed value:", removed_branch)

# Using del
del dict_del["branch"]

print("\nUsing del:")
print(dict_del)

# pop() removes the key and returns its value, whereas del only removes it.

# v. Iterate using .items()

print("\nKey-value pairs:")

for key, value in my_dict.items():
    print(key, "→", value)

# vi. Check whether email exists

if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail key does not exist.")

# vii. Create friend's dictionary

friend_dict = {
    "name": "Riya",
    "roll_no": "1024170400",
    "branch": "COPC",
    "age": 19,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)

# When both dictionaries have the same key, the value from friend_dict wins
# because it is unpacked second.

# viii. Dictionary comprehension for string values only

string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nOnly string values:")
print(string_values)
```
