```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)

max_count = np.max(counts)
most_frequent = values[counts == max_count][0]

indices = np.where(x == most_frequent)[0]

print(most_frequent)
print(indices)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

max_count = np.max(counts)
most_frequent = values[counts == max_count][0]

indices = np.where(y == most_frequent)[0]

print(most_frequent)
print(indices)
```

```python
import numpy as np

# Q2 (a)
array = np.array([
    [1, -2, 3],
    [-4, 5, -6]
])

# i. Element-wise absolute value
print("Absolute value:")
print(np.abs(array))

# ii. 25th, 50th and 75th percentile

print("\nPercentile of flattened array:")
print(np.percentile(array, [25, 50, 75]))

print("\nPercentile for each column:")
print(np.percentile(array, [25, 50, 75], axis=0))

print("\nPercentile for each row:")
print(np.percentile(array, [25, 50, 75], axis=1))

# iii. Mean, Median and Standard Deviation

print("\nMean of flattened array:")
print(np.mean(array))

print("Median of flattened array:")
print(np.median(array))

print("Standard deviation of flattened array:")
print(np.std(array))

print("\nMean of each column:")
print(np.mean(array, axis=0))

print("Median of each column:")
print(np.median(array, axis=0))

print("Standard deviation of each column:")
print(np.std(array, axis=0))

print("\nMean of each row:")
print(np.mean(array, axis=1))

print("Median of each row:")
print(np.median(array, axis=1))

print("Standard deviation of each row:")
print(np.std(array, axis=1))


# Q2 (b)
a = np.array([-1.8, -1.6, -0.5, 0.5, 1.6, 1.8, 3.0])

print("\nFloor:")
print(np.floor(a))

print("\nCeiling:")
print(np.ceil(a))

print("\nTruncated:")
print(np.trunc(a))

print("\nRounded:")
print(np.round(a))
```

```python
import numpy as np

# Q3 (a)
array = np.array([10, 52, 62, 16, 16, 54, 453])

# i. Sorted array
print("Sorted array:")
print(np.sort(array))

# ii. Indices of sorted array
print("\nIndices of sorted array:")
print(np.argsort(array))

# iii. 4 smallest elements
print("\n4 smallest elements:")
print(np.sort(array)[:4])

# iv. 5 largest elements
print("\n5 largest elements:")
print(np.sort(array)[-5:])


# Q3 (b)
array = np.array([1.0, 1.2, 2.2, 2.0, 3.0, 2.0])

# i. Integer elements only
print("\nInteger elements:")
print(array[array % 1 == 0])

# ii. Float elements only
print("\nFloat elements:")
print(array[array % 1 != 0])
```

```python
from PIL import Image
import numpy as np

def img_to_array(path):
    img = Image.open(path)

    # RGB or RGBA image
    if img.mode == "RGB" or img.mode == "RGBA":
        img = img.convert("RGB")
        arr = np.array(img)
        np.savetxt("rgb_image.txt", arr.reshape(-1, 3), fmt="%d")
        print("RGB image saved successfully")

    # Grayscale image
    elif img.mode == "L":
        arr = np.array(img)
        np.savetxt("grayscale_image.txt", arr, fmt="%d")
        print("Grayscale image saved successfully")

    else:
        print("Unsupported image type")
```
